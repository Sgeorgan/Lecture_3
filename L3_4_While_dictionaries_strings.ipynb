## While loops

A `for` loop repeats its code once for every item in a list or `range()`, so we know in advance how many times it will run. Sometimes we do not know that. For example: *how many days does it take for the snow to melt?* A `while` loop repeats its code **as long as a condition is `True`**.

Let's say that there are 42 cm of snow on the ground and that 6.5 cm melt every day:

````python
# Snow depth (cm) at the start, and how much snow melts each day (cm)
snow_depth = 42.0
melt_per_day = 6.5

# We count the days in this variable
day = 0

# Repeat the indented code as long as there is snow on the ground
while snow_depth > 0:
    day = day + 1
    snow_depth = snow_depth - melt_per_day
    print(f"After day {day}: {snow_depth:.1f} cm")

print(f"No snow left after {day} days")
````

This is what Python does:

1. It checks the condition `snow_depth > 0`.
2. If the condition is `True`, it runs the indented code once and goes back to step 1.
3. If the condition is `False`, it skips the indented code and continues with the first line after the loop.

Notice the last value: **-3.5 cm**. The condition is only checked at the *start* of each round. After day 6 there were 3.0 cm left, so the loop ran once more and subtracted the full 6.5 cm. In a real analysis you would report 0 cm, for example with `max(snow_depth, 0)`.

> ⚠️ **Common Pitfall: a `while` loop that never stops**
>
> Something inside the loop must change the condition, otherwise the loop runs forever. In this version we forgot to update `snow_depth`, so `snow_depth > 0` stays `True`:
>
> ```python
> while snow_depth > 0:
>     day = day + 1      # snow_depth never changes - the loop never ends!
> ```
>
> If this happens, stop the code with **Kernel → Interrupt Kernel** in JupyterLab (or the ■ stop button), or with the stop button next to the cell in Google Colab.

> **`for` or `while`?**
>
> | Situation | Use |
> | --- | --- |
> | You want to repeat something once for every item in a list, or a known number of times | `for` |
> | You want to repeat something *until* a condition changes, and you do not know how many rounds that takes | `while` |

## Dictionaries

A list stores values **in order**, and we get a value by its position (index): `cities[0]`. A **dictionary** stores **`key: value` pairs**, and we get a value by its key. This is handy whenever values have a natural name, such as the temperature measured in a city or the season of a month.

We create a dictionary with curly brackets `{}`. Each item is written as `key: value`, and items are separated by commas:

````python
# Temperature (°C) measured at noon in some cities on one day
temperatures = {"Karlstad": 12.5, "Stockholm": 14.1, "Kiruna": 3.2, "Malmo": 15.8}

temperatures
````

We get a value by writing its key in square brackets:

````python
# Look up the value that belongs to the key "Kiruna"
temperatures["Kiruna"]
````

We add a new item, or change an existing one, by assigning a value to a key:

````python
# Add a new key: value pair
temperatures["Lund"] = 15.2

# Change the value of an existing key
temperatures["Karlstad"] = 12.7

temperatures
````

The `in` operator checks whether a **key** exists, and `len()` gives the number of items:

````python
# Is there a key called "Umea" in the dictionary? And how many items are there?
print("Umea" in temperatures)
print(len(temperatures))
````

> ⚠️ **Common Pitfall: `KeyError` for a key that does not exist**
>
> Asking for a key that is not in the dictionary stops your code with an error:
>
> ```python
> temperatures["Umea"]     # KeyError: 'Umea'
> ```
>
> Check first with `if "Umea" in temperatures:`, or use the `.get()` method, which returns a default value instead of an error (see the next cell). Keys are also case-sensitive: `"malmo"` is not the same key as `"Malmo"`.

````python
# .get(key, default) returns the default value when the key does not exist
print(temperatures.get("Umea", "unknown"))
print(temperatures.get("Lund", "unknown"))
````

### Looping over a dictionary

The `.items()` method gives us the key **and** the value of every item, so we can use two loop variables:

````python
# Print each city and its temperature
for city, temp in temperatures.items():
    print(f"{city}: {temp} °C")
````

Dictionaries are also a neat way to *translate* one value into another. In the project, you will need the season of each month. A dictionary with the month number as the key does this in one look-up:

````python
# The season of each month (1 = January, ..., 12 = December)
season_of_month = {
    12: "Winter", 1: "Winter", 2: "Winter",
    3: "Spring", 4: "Spring", 5: "Spring",
    6: "Summer", 7: "Summer", 8: "Summer",
    9: "Autumn", 10: "Autumn", 11: "Autumn",
}

for month in [1, 4, 7, 10, 12]:
    print(f"Month {month} belongs to {season_of_month[month]}")
````

> **Note:** every key in a dictionary is unique. If you write the same key twice, for example `{"Lund": 15.2, "Lund": 15.4}`, the dictionary keeps only the last value. Values, on the other hand, can repeat (three months share the value `"Winter"`).

### Counting with a dictionary

A common task is to count how often each value occurs. We start with an empty dictionary and add 1 to the right key for every observation:

````python
# The weather observed on ten days
observations = ["rain", "sun", "rain", "snow", "sun", "sun", "rain", "fog", "sun", "rain"]

# Start with an empty dictionary
counts = {}

for weather in observations:
    if weather in counts:
        # We have seen this weather type before: add one
        counts[weather] = counts[weather] + 1
    else:
        # First time we see this weather type: start counting at 1
        counts[weather] = 1

print(counts)
````

## String methods

Text (a string) comes with many built-in **methods**: functions that belong to a value and are called with a dot, like `text.upper()`. They are very useful for cleaning up names and for reading information out of text, for example from file names or data files.

````python
# A station name with extra spaces and inconsistent capital letters
name = "  karlstad FLYGPLATS  "

print(name.strip())                                   # remove spaces at the start and end
print(name.strip().title())                           # first letter of each word in upper case
print(name.strip().lower())                           # all letters in lower case
print(name.strip().lower().replace("flygplats", "airport"))   # replace a piece of text
````

As you can see, we can **chain** methods: each method returns a new string, and the next method works on that string.

> ⚠️ **Common Pitfall: string methods do not change the original string**
>
> Strings cannot be changed in place. A method such as `.upper()` returns a **new** string, and the original variable stays as it was:
>
> ```python
> name = "karlstad"
> name.upper()          # creates "KARLSTAD", but we do not store it anywhere
> print(name)           # still prints karlstad
> name = name.upper()   # store the result to keep it
> ```

### Splitting text into pieces

The `.split()` method cuts a string into a **list** of strings at a separator. This is how we can read one line of a data file where the values are separated by semicolons:

````python
# One line of text with a city name and a temperature with its unit (C = Celsius)
record = "Karlstad;-3.5C"

# Cut the text at every semicolon
parts = record.split(";")

print(parts)
print(parts[1])
````

### Indexing, slicing and checking text

Strings can be indexed just like lists: `text[0]` is the first character and `text[-1]` is the last one. A **slice** `text[start:stop]` takes the characters from position `start` up to, but **not including**, position `stop`: `"Karlstad"[0:3]` is `"Kar"`. If you leave out `start` or `stop`, the slice runs from the beginning or to the end, so `text[:-1]` is everything *except* the last character.

The methods `.startswith()` and `.endswith()`, and the `in` operator, check what a string contains.

````python
print("Karlstad"[0:3])            # characters 0, 1 and 2 (the stop position 3 is not included)

temp_text = "-3.5C"

print(temp_text[-1])              # the last character: the unit
print(temp_text[:-1])             # everything except the last character
print(float(temp_text[:-1]))      # ... converted to a number
print(temp_text.endswith("C"))    # True or False
print("." in temp_text)           # does the text contain a full stop?
````

### Putting it together: from text to numbers

Temperatures from different sources are sometimes written with a letter for the unit: C for degrees Celsius and F for degrees Fahrenheit. Let's turn a list of such records into numbers in degrees Celsius, using a `for` loop, `.split()`, slicing, `float()` and `if` statements. We also use `not` to report a value whose unit we do not recognise.

````python
records = ["Karlstad;-3.5C", "Oslo;24.8F", "Stockholm;-1.0C", "Helsinki;23.0X"]

for record in records:
    # Split the record into the city name and the temperature text
    parts = record.split(";")
    name = parts[0]
    temp_text = parts[1]

    # Check that the temperature ends with a unit that we know: C or F
    if not (temp_text.endswith("C") or temp_text.endswith("F")):
        print(f"{name}: cannot read the temperature '{temp_text}'")
        continue

    # The number is everything except the last character
    value = float(temp_text[:-1])

    # Convert Fahrenheit to Celsius
    if temp_text.endswith("F"):
        value = (value - 32) / 1.8

    print(f"{name}: {value:.1f} °C")
````

The `continue` statement skips the rest of the loop body for the current record and goes on with the next one. In Module 4 you will learn how to put code like this into a reusable function.

## Summary

In this lesson you learned three new tools:

- A **`while` loop** repeats code as long as a condition is `True`. Use it when you do not know in advance how many repetitions are needed.
- A **dictionary** stores `key: value` pairs. You look up, add and change values with `d[key]`, check keys with `in`, avoid errors with `.get()`, and loop over the items with `.items()`.
- **String methods** such as `.strip()`, `.lower()`, `.upper()`, `.replace()`, `.split()`, `.startswith()` and `.endswith()` clean up and take apart text. Strings are indexed like lists, and a slice `text[start:stop]` takes a part of a string.

| Watch out for | Reminder |
| --- | --- |
| Endless `while` loop | Something in the loop must change the condition. Stop a running cell with **Interrupt**. |
| `KeyError` | Check with `key in d` first, or use `d.get(key, default)`. Keys are case-sensitive. |
| String methods | They return a **new** string. Store the result: `name = name.strip()`. |
| `.split()` | Gives a **list of strings**. Convert numbers with `float()` or `int()` before calculating. |
| Slices | `text[start:stop]` does **not** include the character at position `stop`. |
